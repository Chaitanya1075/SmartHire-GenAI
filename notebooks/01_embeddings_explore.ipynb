```python
import sys, os
from pathlib import Path
from dotenv import load_dotenv
import google.generativeai as genai
import numpy as np

# notebooks/ is one level below the project root
project_root = Path.cwd().parent
load_dotenv(project_root / ".env")
genai.configure(api_key=os.environ.get("GEMINI_API_KEY"))

print("Project root:", project_root)
```

```python
# Embed a few sample texts to explore how embeddings capture meaning
sample_texts = [
    "Python developer with data analysis experience",
    "Software engineer skilled in backend systems",
    "Professional chef specializing in French cuisine",
]

embeddings = []
for text in sample_texts:
    result = genai.embed_content(
        model="models/gemini-embedding-001",
        content=text,
        task_type="semantic_similarity"
    )
    embeddings.append(result["embedding"])

print(f"Got {len(embeddings)} embeddings, each with {len(embeddings[0])} dimensions")
```

```python
def cosine_similarity(a, b):
    a, b = np.array(a), np.array(b)
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Compare each pair - the two tech-related texts should be much more
# similar to each other than either is to the chef description
for i in range(len(sample_texts)):
    for j in range(i + 1, len(sample_texts)):
        sim = cosine_similarity(embeddings[i], embeddings[j])
        print(f"'{sample_texts[i]}'\nvs\n'{sample_texts[j]}'\nSimilarity: {sim:.4f}\n")
```
